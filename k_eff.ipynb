```python
# Частотность букв русского языка
# Значение N — количество байт, которое занимает символ

frequencies = {
    'о': 0.1099,
    'е': 0.0848,
    'а': 0.0800,
    'и': 0.0737,
    'н': 0.0670,
    'т': 0.0632,
    'с': 0.0547,
    'р': 0.0475,
    'в': 0.0453,
    'л': 0.0434,
    'к': 0.0349,
    'м': 0.0320,
    'д': 0.0298,
    'п': 0.0280,
    'у': 0.0262,
    'я': 0.0200,
    'ы': 0.0190,
    'ь': 0.0174,
    'г': 0.0169,
    'з': 0.0164,
    'б': 0.0159,
    'ч': 0.0145,
    'й': 0.0121,
    'х': 0.0097,
    'ж': 0.0094,
    'ш': 0.0072,
    'ю': 0.0064,
    'ц': 0.0049,
    'щ': 0.0036,
    'э': 0.0033,
    'ф': 0.0027,
    'ъ': 0.0004,
    'ё': 0.0001
}


# ---------------------------------------------------------
# КОДИРОВАНИЕ
# ---------------------------------------------------------

# Сколько байт занимает каждая буква
encoding = {
    'а': 1,
    'б': 1,
    'в': 1,
    'г': 1,
    'д': 1,
    'е': 1,
    'ё': 1,
    'ж': 2,
    'з': 1,
    'и': 1,
    'й': 1,
    'к': 1,
    'л': 1,
    'м': 1,
    'н': 1,
    'о': 1,
    'п': 1,
    'р': 1,
    'с': 1,
    'т': 1,
    'у': 1,
    'ф': 1,
    'х': 2,
    'ц': 2,
    'ч': 2,
    'ш': 2,
    'щ': 2,
    'ъ': 2,
    'ы': 1,
    'ь': 0,
    'э': 1,
    'ю': 2,
    'я': 2
}


# ---------------------------------------------------------
# ДУБЛЕРЫ
# ---------------------------------------------------------

def дублировать(frequencies, duplicates):
    """
    Объединяет частотность дублеров с основными буквами.

    duplicates:
        дублер: основная буква
    """

    frequencies = frequencies.copy()

    for duplicate, original in duplicates.items():

        if duplicate not in frequencies:
            continue

        if original not in frequencies:
            frequencies[original] = 0

        frequencies[original] += frequencies[duplicate]
        del frequencies[duplicate]

    return frequencies


# Какие буквы считаем одинаковыми
duplicates = {
    'ё': 'е',
    'й': 'и'
}


# Применяем объединение
frequencies = дублировать(frequencies, duplicates)


# ---------------------------------------------------------
# РАСЧЁТ
# ---------------------------------------------------------

# Исходный размер:
# UTF-8 в данном случае — 2 байта на кириллический символ
original_weight = sum(
    frequency * 2
    for frequency in frequencies.values()
)

# Новый средний размер символа
new_weight = sum(
    frequency * encoding[letter]
    for letter, frequency in frequencies.items()
)

# Коэффициент сокращения
k_eff = 1 - new_weight / original_weight

# Экономия
saving = k_eff * 100


# ---------------------------------------------------------
# ВЫВОД
# ---------------------------------------------------------

print(f"Исходный средний размер символа: {original_weight:.2f} байт")
print(f"Новый средний размер символа:    {new_weight:.2f} байт")
print()
print(f"k_eff = {k_eff:.2f}")
print(f"Экономия: {saving:.0f}%")
```
